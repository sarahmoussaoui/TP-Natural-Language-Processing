# loading and preprocessing data

```python
import nltk
nltk.download('stopwords')

stop_words = set(stopwords.words('english'))

# function that takes tokens into parameter and returnes them normalized
def normalize_tokens(tokens):
    normalized = []
    for token in tokens:
        token = token.lower()
        token = token.translate(str.maketrans('', '', string.punctuation))
        if token and token not in stop_words:
            normalized.append(token)
    return normalized
```

```python
# current directory
current_dir = os.getcwd()
text_folder = os.path.join(current_dir, "TP/All-in-many")
label_folder = os.path.join(current_dir, "TP/All-in-many_classification")
```

# Probabilities calculations

for each document we need to compute the probability of each class

The most probable class 𝒄" for a given document 𝒅 is computed by selecting the class with the highest product of two probabilities:

o The prior probability of the class 𝑷(𝒄)

o The probability of the document given the class 𝑷(𝒅|𝒄)

              𝒄 = 𝐚𝐫𝐠 𝐦𝐚𝐱 𝑷(𝒄) 𝑷(𝒅|𝒄)
  
o The document 𝒅 can be represented as a set of words:
              𝒄 = 𝐚𝐫𝐠 𝐦𝐚𝐱 𝑷(𝒄) 𝑷(𝒘𝟏, 𝒘𝟐, 𝒘𝟑 … , 𝒘𝒏|𝒄)

o The probability 𝑷(𝒄) represents the proportion of documents in the training set that belong to class c
              
              𝑷(𝒄) = 𝑵𝒄 / 𝑵


𝑵𝒄 is the number of documents in the training set belonging to class 𝒄.

𝑵 is the total number of documents in the training set.

o The probability 𝑷(𝒘𝟏, 𝒘𝟐, 𝒘𝟑 … , 𝒘𝒏|𝒄) is computed as the product of the individual probabilities:

            𝑷(𝒘𝟏, 𝒘𝟐, 𝒘𝟑 … , 𝒘𝒏|𝒄) = product 𝑷(𝒘𝒊|𝒄) i ranging from 1 to n

Where:

The probability 𝑷(𝒘𝒊|𝒄) corresponds to the relative frequency of the word 𝒘𝒊 among all words in documents belonging to class 𝒄:

              𝑷(𝒘𝒊|𝒄) = (𝐜𝐨𝐮𝐧𝐭(𝒘𝒊, 𝒄) + 𝟏) / ∑ 𝐜𝐨𝐮𝐧𝐭(𝒘, 𝒄) + 𝑽
              
Where:
𝑽 represents the vocabulary of the training set.

in the interface we are meant to show: 
- P(c) : estimated class probabilities for a selected article
- P(w/c): estimated class probabilities for each word in the article

```python
# let's create a dictionnary formed like:
# data = {
#   "article_number":..
#    "label" :  ..
#    "text": ..
#    "tokens" : ..
#    "normalized_tokens with porter stemmer"  : ..
#    "P(normalize token | label)" : [value for class1, class2, class3, class 4] each token probability given the label
#    "P(label)" : [value for class1, class2, class3, class 4] prior probability of each label
# }

data = []
for filename in os.listdir(text_folder):
    if filename.endswith(".txt"):
        article_number = filename[8] # its written like Article_1_Volume_18
        with open(os.path.join(text_folder, filename), 'r', encoding='utf-8') as f:
            text = f.read()
            tokens = text.split()
            # porter stemmer normalization
            normalized_tokens = normalize_tokens(tokens)
        with open(os.path.join(label_folder, filename), 'r', encoding='utf-8') as f:
            label = f.read()
        data.append({
            "article_number": article_number,
            "label": label,
            "text": text,
            "tokens": tokens,
            "normalized_tokens": normalized_tokens,
        })
```

```python
# saving data dictionary to a file for later use
import pickle
with open('data.pkl', 'wb') as f:
    pickle.dump(data, f)
```
